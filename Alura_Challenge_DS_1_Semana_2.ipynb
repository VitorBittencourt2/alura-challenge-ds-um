Imports

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Leitura do arquivo

```python
dados = pd.read_json("/content/drive/MyDrive/Colab Notebooks/Alura Challenge DS 1/Telco-Customer-Churn-limpeza.json")
dados.head()
```

Distribuição do Churn

```python
plt.figure(figsize=(8, 6))
churn_counts = dados['Churn'].value_counts()

ax = sns.barplot(x=churn_counts.index, y=churn_counts.values, palette='viridis')
plt.title('Distribuição da variável Churn')
plt.xlabel('Churn')
plt.ylabel('Quantidade')

# Adicionando os números em cima das barras
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=12)

plt.show()
```

```python
plt.figure(figsize=(10, 6))
churn_gender_counts = dados.groupby(['gender', 'Churn']).size().reset_index(name='Counts')

# Criando o gráfico com distinção de churn por gênero
ax = sns.barplot(data=churn_gender_counts, x='gender', y='Counts', hue='Churn', palette='viridis')

# Ajustando o título e os rótulos
plt.title('Distribuição da variável gênero por Churn', fontsize=16)
plt.xlabel('Gender', fontsize=14)
plt.ylabel('Quantidade', fontsize=14)

# Adicionando números em cima das barras
for container in ax.containers:
    ax.bar_label(container, fmt='%d', label_type='edge', fontsize=12)

plt.show()
```

```python
columns = ['tenure', 'Charges.Monthly']

for col in columns:
    plt.figure(figsize=(10, 6))
    sns.boxplot(data=dados, x=col, y='Churn', palette='viridis', orient='h')  # Orientação horizontal
    plt.title(f'Distribuição de {col} por Churn', fontsize=16)
    plt.xlabel(col, fontsize=14)
    plt.ylabel('Churn', fontsize=14)
    plt.show()
```

```python
numerical_data = dados.select_dtypes(include=['number'])
correlation_matrix = numerical_data.corr()


plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)


plt.title('Matriz de Correlação')
plt.show()
```
